## Célula 2 — Criação da Estrutura de Pastas

````python
from pathlib import Path

BASE = Path(PASTA_PROJETO)

PASTAS = [
    'data/externos',
    'data/originais',
    'data/pre-tratados',
    'data/logistics',
    'data/finance',
    'notebooks/vendas',
    'notebooks/logistica',
    'notebooks/financas',
    'exports',
    'summary_logs',
]

print('Criando estrutura de pastas...\n')

for pasta in PASTAS:
    caminho = BASE / pasta
    caminho.mkdir(parents=True, exist_ok=True)
    # Cria .gitkeep para que pastas vazias sejam versionadas
    gitkeep = caminho / '.gitkeep'
    if not gitkeep.exists():
        gitkeep.touch()
    print(f'  ✅ {pasta}/')

print('\nEstrutura de pastas criada com sucesso.')
````

## Célula 5 — README.md

````python
readme = f"""\
# Projeto Fictus
### Due Diligence de Aquisição — Análise completa de um e-commerce via dados

![Status](https://img.shields.io/badge/status-concluído-brightgreen)
![Python](https://img.shields.io/badge/python-3.10+-blue)
![Dataset](https://img.shields.io/badge/dataset-Olist-orange)
![Notebooks](https://img.shields.io/badge/notebooks-18-lightgrey)

---

## O que é o Projeto Fictus

O Projeto Fictus simula uma **due diligence completa de aquisição** de uma empresa de e-commerce fictícia.
A partir do dataset público Olist (2016–2018, deslocado +7 anos para o período 2023–2025), três frentes
analíticas independentes — Vendas, Logística e Finanças — convergem para um **Veredito Estratégico**
gerado por IA.

O objetivo não é apenas analisar dados. É responder uma pergunta de negócio real:
**essa empresa vale ser comprada?**

---

## Estrutura do Repositório

```
fictus/
├── README.md
├── requirements.txt
├── .gitignore
├── data/
│   ├── externos/            ← arquivos baixados do IBGE
│   ├── originais/           ← CSVs originais do Olist (não versionados)
│   ├── pre-tratados/        ← gerado pelo ETL de Vendas
│   ├── logistics/           ← gerado pelo ETL de Logística
│   └── finance/             ← gerado pelo ETL de Finanças
├── notebooks/
│   ├── vendas/
│   ├── logistica/
│   └── financas/
├── exports/                 ← gráficos gerados (.png)
├── summary_logs/            ← logs gerados ao final de cada frente analítica
└── fictus_veredito_estrategico.md
```

---

## As Três Frentes Analíticas

### Frente 1 — Análise de Vendas
> **A empresa se financia ou consome capital à medida que cresce?**

11 análises em 5 blocos: viabilidade econômica, motor de receita, escalabilidade operacional,
riscos ocultos e cenários de sobrevivência.

### Frente 2 — Análise Logística
> **O modelo terceirizado chegou no seu limite — e quanto está custando isso?**

Diagnóstico do modelo atual, análise make-or-buy, impacto do SLA na satisfação,
escalabilidade sob picos e custo de reversão.

### Frente 3 — Análise Financeira
> **Quem está financiando o crescimento — a empresa ou os intermediários?**

Perfil de pagamentos, risco de crédito latente, spread líquido, escalabilidade
da estrutura de recebimento e cenários de reestruturação.

---

## Como Rodar o Projeto

### 1. Instalar dependências
```bash
pip install -r requirements.txt
```

### 2. Baixar o dataset
Baixe o [Olist E-Commerce Public Dataset](https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce)
no Kaggle e salve os CSVs em `data/originais/`.

### 3. Ordem de execução
```
1. notebooks/vendas/00_ETL_analise_vendas.ipynb
2. notebooks/vendas/01 ao 05
3. notebooks/logistica/00_ETL_analise_logistica.ipynb
4. notebooks/logistica/01 ao 05
5. notebooks/financas/00_ETL_analise_financeira.ipynb
6. notebooks/financas/01 ao 05
```

Os Summary Logs gerados ao final de cada frente ficam salvos automaticamente em `summary_logs/`.

---

## Como Usar o Veredito Estratégico

1. Execute todos os notebooks até o final
2. Abra os arquivos em `summary_logs/` (gerados automaticamente)
3. Abra `fictus_veredito_estrategico.md`
4. Escolha o perfil de tese: Conservadora, Moderada ou Arrojada
5. Cole os três Summary Logs nos campos indicados
6. Cole o prompt completo em Claude, ChatGPT ou Gemini
7. Receba o parecer executivo completo

---

## Stack Tecnológica

| Camada | Tecnologia |
|---|---|
| Linguagem | Python 3.10+ |
| Manipulação de dados | pandas, numpy |
| Visualização | matplotlib, seaborn |
| Estatística | scipy |
| Ambiente | Jupyter Lab |
| Veredito final | Claude / ChatGPT / Gemini via prompt |
| Dataset | Olist E-Commerce Public Dataset (Kaggle) |

---

## Sobre o Autor

Em transição de carreira para a área de dados. O Projeto Fictus é o primeiro projeto de portfólio.

[![LinkedIn](https://img.shields.io/badge/LinkedIn-conectar-0077B5?logo=linkedin)]({AUTOR_LINKEDIN})

---

## Licença

MIT License — livre para uso, estudo e adaptação com atribuição.
"""

arquivo = BASE / 'README.md'
arquivo.write_text(readme.strip(), encoding='utf-8')

print('✅ README.md criado')
````
